# **LLM & Agentic AI Fundamentals**

##Tokens
Tokens are the units into which text is divided by a tokenizer before being processed by an LLM. They can represent words, parts of words, punctuation, or characters, and the model processes their corresponding numerical IDs.


###Context Window

The context window is the maximum number of tokens an LLM can process in a single interaction, including the input context and the generated output.

# Transformer — Quick Notes

* **Transformer:** Neural network architecture based on **Attention**, used by modern LLMs.
* **Purpose:** Understand relationships between different tokens in a sequence.
* **Basic flow:** `Text → Tokens → Token IDs → Embeddings → Positional Information → Transformer Blocks → Output`
* **Tokenization:** Converts text into tokens (words, subwords, punctuation, etc.).
* **Embedding:** Converts token IDs into numerical vectors.
* **Positional Encoding:** Provides token order/position; modern LLMs commonly use **RoPE**.
* **Self-Attention:** Determines which tokens are important to each other.
* **Q (Query):** What information am I looking for?
* **K (Key):** What information do I contain?
* **V (Value):** What information should I provide?
* **Multi-Head Attention:** Multiple attention heads learn different relationships.
* **FFN (Feed-Forward Network):** Further processes the representation after attention.
* **Residual Connection:** Adds the original input to the layer output.
* **Layer Normalization:** Helps stabilize the network.
* **Transformer Block:** Attention + FFN + Residual Connections + Normalization.
* **Causal Attention:** In decoder LLMs, a token cannot see future tokens.
* **Decoder-only:** Used by most modern generative LLMs (GPT-style, Llama, etc.).
* **Next-Token Prediction:** LLM predicts one token at a time to generate the response.
* **Context Window:** Maximum number of tokens the LLM can process in one interaction.
* **RAG + Transformer:** `Retrieve → Rerank → Context → Prompt → LLM → Answer`
* **Key idea:** **Transformer = Attention + Neural Network layers → understand relationships between tokens and generate/process information.**


# Inference — Quick Notes

* **Inference:** The process of using a **trained AI/ML model to make predictions or generate outputs** for new input data.
* In GenAI, inference means **sending a prompt/input to an LLM and getting its generated output**.
* **Training:** Model learns from data → updates weights.
* **Inference:** Model uses learned weights → generates output; **weights are not updated**.
* **LLM inference flow:** `Prompt → Tokenization → Transformer → Next-Token Prediction → Generated Tokens → Text`
* **Autoregressive inference:** The LLM generates one token, adds it to the context, and predicts the next token repeatedly.
* **Inference parameters:** Temperature, Top-P, Max Output Tokens, etc. can influence generation.
* **Inference latency:** Time required to generate the response.
* **Inference cost:** Computational/API cost of running the model.
* **Key idea:** **Training teaches the model; inference uses what it learned.**


# Temperature & Top-P — Quick Notes

* **Temperature:** Controls how random/creative the LLM's token selection is.

* **Low temperature (e.g. 0–0.3):** More deterministic, consistent, focused output.

* **High temperature (e.g. 0.7–1.5):** More diverse, creative, unpredictable output.

* **Example:** `Temperature = 0` → usually chooses the highest-probability token; `Temperature = 1` → more variation.

* **Top-P (Nucleus Sampling):** Controls token selection by considering only the smallest group of most-probable tokens whose combined probability reaches **P**.

* **Low Top-P:** Smaller candidate pool → more focused output.

* **High Top-P:** Larger candidate pool → more diverse output.

* **Example:** `Top-P = 0.9` → select from the smallest set of tokens whose cumulative probability is ≥ 90%.

### Difference

* **Temperature →** controls **how randomly** tokens are selected.
* **Top-P →** controls **how many probable tokens** are considered.
* Usually, **adjust one rather than aggressively tuning both together**.

### Simple Example

```text
Prompt → "Write a story about a dog"

Temperature ↓ → predictable / consistent story
Temperature ↑ → creative / varied story

Top-P ↓ → choose from fewer likely tokens
Top-P ↑ → consider more likely tokens
```

### Production Rule

* **RAG / factual tasks:** Lower temperature → more consistent responses.
* **Creative tasks:** Higher temperature → more diverse responses.
* **Important:** Exact behavior and supported ranges depend on the specific model/API.

> **Temperature controls randomness; Top-P controls the probability-based candidate pool.**

### Agent Loop

An agent loop is the repeated cycle where an AI agent observes the current state, reasons about what to do, takes an action such as calling a tool, observes the result, and repeats until the goal is completed.

**Agent Loop = Think → Act → Observe → Update State → Repeat → Finish**

### State

State is the structured information an agent maintains and updates throughout its workflow so that each step knows what has already happened and what information is currently available.

###Planning

# Planning & Agent Loop

## 1. Planning

- **What:** Planning is the process where an AI agent decides the **steps/actions required to achieve a goal**.

- **When to use:**
  - Complex tasks with multiple steps.
  - Tasks with dependencies between steps.
  - When having a clear sequence improves reliability.

- **How:**
  `Goal → Create Plan → Step 1 → Step 2 → Step 3 → Final Result`

- **Example:**
  User: "My ₹5,000 payment failed but money was deducted. Please resolve it."

  Agent Plan:
  1. Check transaction.
  2. Check payment status.
  3. Check whether money was deducted.
  4. Raise complaint if required.
  5. Inform the user.

- **Advantages:**
  - Clear and structured workflow.
  - Easier to debug.
  - More predictable execution.
  - Useful for complex multi-step tasks.

- **Disadvantages:**
  - Requires additional LLM calls/tokens.
  - Adds latency.
  - A fixed plan can become invalid when new information appears.
  - Unnecessary for simple tasks.

- **Remember:**
  `Planning = "What should I do?"`

---

## 2. Agent Loop

- **What:** An Agent Loop is a continuous cycle where the agent **decides → acts → observes the result → updates state → decides again** until the goal is completed.

- **When to use:**
  - When the next action depends on the previous result.
  - When tools are involved.
  - When the task has uncertain or changing outcomes.
  - When the agent may need to re-plan dynamically.

- **How:**
  `Goal → Decide → Act/Tool Call → Observe Result → Update State → Decide Again → Repeat → Final Answer`

- **Example:**
  User: "My ₹5,000 payment failed but money was deducted."

  Agent Loop:
  1. Decide → Check transaction.
  2. Observe → Transaction found.
  3. Decide → Check payment status.
  4. Observe → Payment failed.
  5. Decide → Check deduction.
  6. Observe → Money deducted.
  7. Decide → Raise complaint.
  8. Observe → Complaint created.
  9. Final Answer → Inform user.

- **Advantages:**
  - Flexible and adaptive.
  - Reacts to tool results.
  - Can dynamically change the next action.
  - Useful for complex and uncertain tasks.

- **Disadvantages:**
  - Multiple LLM calls increase cost.
  - Increases latency.
  - Agent can get stuck in loops.
  - Less predictable than a fixed workflow.

- **Remember:**
  `Agent Loop = Decide → Act → Observe → Decide Again`

---

## Planning vs Agent Loop

| Planning | Agent Loop |
|---|---|
| Creates a roadmap | Executes and adapts |
| "What should I do?" | "What should I do next?" |
| Can create steps upfront | Repeated decision-making |
| More predictable | More flexible |
| Useful for complex planned tasks | Useful when results affect the next action |

### Key Idea

`Planning = Create the roadmap`

`Agent Loop = Execute → Observe → Adapt`

Planning and Agent Loop can work together, but an agent does **not always need a separate planning phase**. For simple tasks, the agent can directly use the loop.

###Tool Calling

Tool calling is the mechanism that allows an LLM/agent to request the execution of an external function or tool when it needs information or an action that the model itself cannot perform reliably.

The important point:

The LLM does not directly execute the tool. It decides which tool to call and provides the required arguments; your application executes the tool and returns the result to the LLM.

# Tool Selection

- **What:** Tool Selection is the process where an agent decides **which tool to use** to complete the current task.

- **When to use:**
  - When an agent has multiple tools available.
  - When different tools perform different actions.
  - When the next tool depends on the user's request or previous result.

- **How:**
  `User Goal → Agent/LLM → Select Tool → Tool Call → Tool Result → Agent`

- **Example:**
  User: "My ₹5,000 payment failed but money was deducted."

  Available tools:
  - `get_transaction()`
  - `get_payment_status()`
  - `create_complaint()`
  - `refund_payment()`

  Agent:
  1. Select `get_transaction()`
  2. Select `get_payment_status()`
  3. Based on result, select `create_complaint()`

- **Tool Selection vs Tool Calling:**
  - Tool Selection = **Which tool should I use?**
  - Tool Calling = **Call the selected tool with arguments.**

- **Advantages:**
  - Enables agents to work with multiple tools.
  - Makes workflows dynamic.
  - Helps choose the appropriate action based on context.

- **Disadvantages:**
  - Too many tools can confuse the model.
  - Poor tool descriptions can cause wrong selection.
  - Incorrect tool selection can lead to wrong or unsafe actions.

- **Best Practices:**
  - Use clear tool names and descriptions.
  - Keep tools focused and specific.
  - Use structured arguments.
  - Validate permissions before executing sensitive tools.

- **Remember:**
  `Tool Selection = Choose the tool`
  `Tool Calling = Execute the chosen tool`

# Conditional Routing

- **What:** Conditional Routing is the process where an agent/workflow **chooses the next step based on the current state, input, or tool result**.

- **When to use:**
  - When different conditions require different actions.
  - When the next step depends on the previous result.
  - When building dynamic agent workflows.
  - Useful for classification, validation, approval, error handling, and multi-agent systems.

- **How:**
  `Input → Agent/Node → Evaluate Condition → Choose Route → Next Node`

- **Example:**
  User: "My payment failed."

  Agent checks payment status:

  `Payment Status`
  - `SUCCESS` → Inform user
  - `FAILED` → Check deduction
  - `PENDING` → Ask user to wait
  - `UNKNOWN` → Retry/check system

- **Simple Flow:**

  `Check Status`
       ↓
  `Condition`
       ├── SUCCESS → Final Answer
       ├── FAILED → Check Deduction
       ├── PENDING → Wait/Inform
       └── UNKNOWN → Retry

- **Advantages:**
  - Makes workflows dynamic.
  - Supports different paths for different situations.
  - Improves control and reliability.
  - Easy to combine with LangGraph.

- **Disadvantages:**
  - Too many conditions can make workflows complex.
  - Incorrect conditions can send the agent down the wrong path.
  - Large routing logic can become difficult to maintain.

- **Best Practices:**
  - Keep routing conditions simple and explicit.
  - Use structured state values.
  - Handle unknown/unexpected cases.
  - Keep business-critical routing deterministic when possible.

- **Remember:**
  `Conditional Routing = Decide WHERE to go next based on a condition`

- **Planning vs Conditional Routing:**
  - Planning = **What steps are needed?**
  - Conditional Routing = **Which path should I take now?**
  - Agent Loop = **Decide → Act → Observe → Decide Again**

# Multi-Step Workflows

- **What:** A Multi-Step Workflow is a process where an agent completes a task through **multiple connected steps**, where the output of one step can become the input for the next step.

- **When to use:**
  - When a task requires multiple actions.
  - When steps have dependencies.
  - When a simple single LLM call is not enough.
  - Common in real-world business and AI systems.

- **How:**
  `Goal → Step 1 → Step 2 → Step 3 → Step 4 → Final Result`

- **Real-Life Example 1 — Payment Complaint:**
  User: "₹5,000 was deducted but my payment failed."

  `User Request`
       ↓
  `Check Transaction`
       ↓
  `Check Payment Status`
       ↓
  `Check Money Deduction`
       ↓
  `Create Complaint`
       ↓
  `Notify User`

- **Real-Life Example 2 — Customer Support:**
  User: "My order hasn't arrived."

  `Identify Customer`
       ↓
  `Find Order`
       ↓
  `Check Delivery Status`
       ↓
  `Identify Problem`
       ↓
  `Take Action`
       ↓
  `Respond to Customer`

- **Real-Life Example 3 — RAG System:**
  User: "What is our company's refund policy?"

  `User Question`
       ↓
  `Query Understanding`
       ↓
  `Retrieve Documents`
       ↓
  `Rerank Results`
       ↓
  `Generate Answer`
       ↓
  `Validate Answer`
       ↓
  `Final Response`

- **Advantages:**
  - Breaks complex tasks into manageable steps.
  - Easier to debug and monitor.
  - Each step can have a specific responsibility.
  - Can combine LLMs, tools, APIs, databases, and business logic.

- **Disadvantages:**
  - More steps can increase latency and cost.
  - Failure in one step can affect later steps.
  - Complex workflows can become difficult to maintain.

- **Multi-Step Workflow vs Agent Loop:**
  - Multi-Step Workflow = **Multiple connected steps to complete a task.**
  - Agent Loop = **Decide → Act → Observe → Decide again.**
  - A multi-step workflow can contain an agent loop.

- **Remember:**
  `Multi-Step Workflow = Break one complex task into multiple connected steps.`

| Concept                 | Main Question                                            | Purpose                                        |
| ----------------------- | -------------------------------------------------------- | ---------------------------------------------- |
| **Planning**            | **What should I do?**                                    | Creates a plan/roadmap                         |
| **Multi-Step Workflow** | **What steps make up this task?**                        | Defines the complete sequence of work          |
| **Conditional Routing** | **Which path should I take?**                            | Chooses between multiple paths                 |
| **Agent Loop**          | **What should I do next based on what I just observed?** | Repeatedly decides, acts, observes, and adapts |


# Memory

- **What:** Memory is the mechanism that allows an AI agent to **retain information and use it in future interactions or tasks**.

- **When to use:**
  - When the agent needs information from previous conversations.
  - When maintaining user preferences or history.
  - When a task continues across multiple sessions.
  - Useful for personalized AI assistants and long-running agents.

- **Types:**
  - **Short-Term Memory:** Information available during the current conversation/workflow.
  - **Long-Term Memory:** Information stored and retrieved across future conversations or executions.

- **How:**
  `User → Agent → Store Important Information → Future Request → Retrieve Memory → Agent`

- **Real-Life Example — Banking Assistant:**
  User: "My preferred language is Hindi."

  Agent stores:
  `language = Hindi`

  Later:

  User: "Explain my transaction."

  Agent retrieves the preference and responds in Hindi.

- **Real-Life Example — Coding Assistant:**
  User: "I use FastAPI and PostgreSQL for my backend."

  Later:
  "Create an API for my project."

  Agent retrieves the previous preference and can suggest a FastAPI + PostgreSQL solution.

- **State vs Memory:**
  - **State:** Information needed during the current workflow.
  - **Memory:** Information retained for future interactions.

- **Advantages:**
  - Personalization.
  - Continuity across conversations.
  - Better long-running task support.
  - Reduces the need for users to repeat information.

- **Disadvantages:**
  - Incorrect memories can lead to incorrect responses.
  - Storage and retrieval add complexity.
  - Privacy and security must be considered.
  - Unnecessary information should not be stored.

- **Remember:**
  `State = Current Workflow`
  
  `Memory = Future/Previous Interactions`

  `Memory = Store → Retrieve → Use`

# Reflection

- **What:** Reflection is the process where an AI agent **reviews its own output or actions, identifies problems, and improves the result**.

- **When to use:**
  - When output quality is important.
  - When the first answer may contain mistakes.
  - For code generation, reasoning, writing, RAG, and complex tasks.
  - When validation or self-correction is required.

- **How:**
  `Task → Generate Output → Evaluate Output → Identify Problems → Improve → Final Output`

- **Real-Life Example — Coding Agent:**
  User: "Create a Python API for user registration."

  Agent:
  1. Generate FastAPI code.
  2. Review the code.
  3. Detect missing input validation.
  4. Fix the validation.
  5. Return improved code.

- **Real-Life Example — RAG:**
  User: "What is our refund policy?"

  Agent:
  1. Retrieve relevant documents.
  2. Generate answer.
  3. Check whether the answer is supported by retrieved documents.
  4. Detect unsupported information.
  5. Correct the answer.
  6. Return final answer.

- **Simple Flow:**

  `Generate → Evaluate → Improve → Generate/Final`

- **Advantages:**
  - Improves output quality.
  - Can detect and correct errors.
  - Useful for complex tasks.
  - Adds an additional quality-control layer.

- **Disadvantages:**
  - Requires additional LLM calls.
  - Increases cost and latency.
  - The agent can sometimes incorrectly criticize a correct answer.
  - Reflection does not guarantee correctness.

- **Reflection vs Agent Loop:**
  - **Agent Loop:** Decides what action to take next.
  - **Reflection:** Reviews whether the result/action was good and improves it.

- **Remember:**
  `Reflection = Generate → Review → Improve`

# Retry / Recovery

- **What:** Retry / Recovery is the mechanism that allows an AI agent to **handle failures and try again or take an alternative action** instead of completely failing.

- **When to use:**
  - API/tool failures.
  - Temporary network errors.
  - LLM failures or timeouts.
  - Invalid tool arguments.
  - External service temporarily unavailable.
  - When an alternative path can recover the task.

- **How:**
  `Action → Failure → Retry / Alternative Action → Success → Continue`

- **Real-Life Example — Payment API:**
  User: "Check my payment status."

  `Agent`
      ↓
  `Payment API`
      ↓
  `Timeout`
      ↓
  `Retry`
      ↓
  `Payment API`
      ↓
  `Success`
      ↓
  `Return Status`

- **Recovery Example:**

  `Primary API`
       ↓
    Failure
       ↓
  `Retry 1`
       ↓
    Failure
       ↓
  `Retry 2`
       ↓
    Failure
       ↓
  `Fallback API`
       ↓
    Success

- **Important:** Retry should **not continue forever**.

  Example:
  `Maximum retries = 3`

  After 3 failures:
  - Use fallback.
  - Ask for human intervention.
  - Return a controlled error.

- **Common Strategies:**
  - **Immediate Retry:** Try the same operation again.
  - **Exponential Backoff:** Wait longer between retries.
  - **Fallback:** Use another tool/service.
  - **Recovery:** Correct the input/state and try again.
  - **Human Escalation:** Ask a human when automatic recovery fails.

- **Advantages:**
  - Improves reliability.
  - Handles temporary failures.
  - Prevents the entire workflow from failing.
  - Useful for production AI systems.

- **Disadvantages:**
  - Additional latency.
  - Additional API/LLM cost.
  - Repeated actions can be dangerous for non-idempotent operations.
  - Poor retry logic can create infinite loops.

- **Retry vs Recovery:**
  - **Retry:** Try the same operation again.
  - **Recovery:** Take corrective or alternative action after failure.

- **Remember:**
  `Retry = Try Again`
  
  `Recovery = Fix/Fallback and Continue`

  `Failure ≠ Final Failure`

# Human-in-the-Loop (HITL)

- **What:** Human-in-the-Loop is a mechanism where an AI agent **pauses its workflow and asks a human for approval, input, or intervention** before continuing.

- **When to use:**
  - High-risk actions.
  - Financial transactions.
  - Sensitive data operations.
  - Irreversible actions.
  - When the agent is uncertain.
  - When human judgment is required.

- **How:**
  `Agent → Analyze → Human Approval/Input → Continue → Final Result`

- **Real-Life Example — Refund:**
  User: "Refund my ₹50,000 payment."

  `Agent`
      ↓
  `Check Transaction`
      ↓
  `Check Refund Eligibility`
      ↓
  `Refund Amount = ₹50,000`
      ↓
  `Human Approval`
      ↓
  `Approved`
      ↓
  `Refund API`
      ↓
  `Success`

- **If Human Rejects:**

  `Agent → Human Review → Rejected → Stop/Alternative Action`

- **Another Example — Customer Support:**
  - Agent detects a potentially fraudulent transaction.
  - Agent gathers transaction details.
  - Agent pauses.
  - Human fraud analyst reviews it.
  - Human approves/rejects the action.
  - Agent continues accordingly.

- **Types:**
  - **Approval:** Human approves/rejects an action.
  - **Input:** Human provides missing information.
  - **Escalation:** Agent transfers a difficult case to a human.
  - **Review:** Human checks the agent's generated result.

- **Advantages:**
  - Improves safety.
  - Prevents dangerous automated actions.
  - Adds human judgment where needed.
  - Useful for high-stakes systems.

- **Disadvantages:**
  - Slower than fully automated workflows.
  - Requires human availability.
  - Can reduce scalability.
  - Poorly designed HITL can create unnecessary interruptions.

- **HITL vs Reflection:**
  - **Reflection:** AI reviews its own output.
  - **HITL:** Human reviews/approves the AI's output or action.

- **Remember:**
  `HITL = AI works → Human intervenes when needed → AI continues`

# Agent Evaluation

- **What:** Agent Evaluation is the process of **measuring how well an AI agent performs its task** using defined metrics, test cases, and expected outcomes.

- **When to use:**
  - Before deploying an agent to production.
  - After changing the model, prompt, tools, or workflow.
  - To compare different agents/models.
  - For continuous production monitoring.

- **How:**
  `Test Dataset → Run Agent → Collect Results → Evaluate Metrics → Analyze Failures → Improve → Re-evaluate`

- **Real-Life Example — Customer Support Agent:**

  Test:
  `User: "My payment failed but ₹5,000 was deducted."`

  Expected:
  - Correctly identify the issue.
  - Check the correct transaction.
  - Use the correct tools.
  - Create complaint when required.
  - Give an accurate response.

  Evaluate:
  `Correctness + Tool Selection + Tool Arguments + Response Quality + Latency + Cost`

- **Important Evaluation Areas:**
  - **Task Success:** Did the agent complete the task?
  - **Answer Quality:** Is the final answer correct?
  - **Tool Selection:** Did it choose the correct tool?
  - **Tool Arguments:** Were the parameters correct?
  - **Groundedness:** Is the answer supported by available data?
  - **Latency:** How long did it take?
  - **Cost:** How many tokens/API calls were used?
  - **Safety:** Did it avoid unsafe or unauthorized actions?

- **RAG Example:**
  `Question → Retrieve → Rerank → Generate → Evaluate`

  Possible metrics:
  - Context Precision
  - Context Recall
  - Faithfulness
  - Answer Relevance

- **Agent Evaluation Flow:**

  `Test Cases`
       ↓
  `Run Agent`
       ↓
  `Trace Agent Execution`
       ↓
  `Evaluate`
       ↓
  `Find Failure`
       ↓
  `Improve`
       ↓
  `Run Again`

- **Advantages:**
  - Makes agent quality measurable.
  - Helps detect failures.
  - Enables model/workflow comparison.
  - Important for production reliability.

- **Disadvantages:**
  - Creating good evaluation datasets takes effort.
  - Some AI outputs are subjective.
  - Evaluation itself can add cost.
  - Good benchmark results do not guarantee perfect production behavior.

- **Remember:**
  `Agent Evaluation = Measure → Find Failures → Improve → Measure Again`

- **Production Principle:**
  `Don't ask "Does the agent work?"`
  
  `Ask "How well does it work, where does it fail, and how often?"`
